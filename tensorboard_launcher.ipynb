```python
import threading, time, shutil, os

LOCAL_LOG_DIR = "/local_disk0/runs/"
DBFS_LOG_DIR = "/Volumes/personal_schema/kschlachter/v_kschlachter/diamond/runs/"
os.makedirs(LOCAL_LOG_DIR, exist_ok=True)

def sync_logs():
    try:
        print("Syncing logs to volumes")
        shutil.copytree(LOCAL_LOG_DIR, DBFS_LOG_DIR, dirs_exist_ok=True)
    except Exception as e:
        print(f"Log sync error: {e}")

def periodic_sync(interval=300):  # every 5 minutes
    while True:
        sync_logs()
        time.sleep(interval)

# Start sync thread
t = threading.Thread(target=periodic_sync, args=(300,), daemon=True)
t.start()
```
